```python
import tensorflow as tf 
from tensorflow import keras 
from keras.layers import Dense,Conv2D,MaxPooling2D,Flatten
from keras.models import Sequential
```

```python
# generator

train_dataset=keras.utils.image_dataset_from_directory(
    directory=r"D:\COURSES\AIML\DEEP LEARNING\PART-2\Projects\train",
    labels="inferred",
    label_mode="int",
    batch_size=32,
    image_size=(256,256)
)


test_dataset=keras.utils.image_dataset_from_directory(
    directory=r"D:\COURSES\AIML\DEEP LEARNING\PART-2\Projects\test",
    labels="inferred",
    label_mode="int",
    batch_size=32,
    image_size=(256,256)
)
```

```python
train_dataset
```

```python
def process(image, label):
    image = tf.cast(image / 255., tf.float32)
    return image, label

train_datasets = train_dataset.map(process)
test_dataset = test_dataset.map(process)
```

```python
model=Sequential([
    Conv2D(32,kernel_size=(3,3),activation="relu",padding="valid",input_shape=(256,256,3)),
    MaxPooling2D(pool_size=(2,2),strides=2,padding="valid"),
    Conv2D(64,kernel_size=(3,3),activation="relu",padding="valid"),
    MaxPooling2D(pool_size=(2,2),strides=2,padding="valid"),
    Conv2D(64,kernel_size=(3,3),activation="relu",padding="valid"),
    MaxPooling2D(pool_size=(2,2),strides=2,padding="valid"),
    Flatten(),
    Dense(128,activation="relu"),
    Dense(64,activation="relu"),
    Dense(1,activation="sigmoid")
])
```

```python
model.summary()
```

```python
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
```

```python
history=model.fit(train_dataset,
                 epochs=1,
                 batch_size=10,
                 verbose=1)
```

```python
loss,acc=model.evaluate(test_dataset,verbose=1)
```
